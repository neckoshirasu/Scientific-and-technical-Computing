# ニュートン法（多次元 $n=m$）


方程式$\boldsymbol{f}(\boldsymbol{x})=\boldsymbol{0}$を現在の解$\boldsymbol{x}^{(k)}$の周りで1次テーラー近似し，
整理すると

- $
\boldsymbol{x}^{(k+1)} = \boldsymbol{x}^{(k)} - J(\boldsymbol{x}^{(k)})^{-1} \boldsymbol{f}(\boldsymbol{x}^{(k)})
$

ここでニュートンステップを $\boldsymbol{d}^{(k)} = \boldsymbol{x}^{(k+1)} - \boldsymbol{x}^{(k)}$ とおくと

- $
J(\boldsymbol{x}^{(k)}) \boldsymbol{d}^{(k)}
= \boldsymbol{f}(\boldsymbol{x}^{(k)})
$ を解く
- $\boldsymbol{x}^{(k+1)} = \boldsymbol{x}^{(k)} +\boldsymbol{d}^{(k)}$


## ニュートン法の問題点

- ヤコビ行列 $J(\boldsymbol{x}^{(k)})$ の計算コストが高い
- ヤコビ行列の条件数が大きい場合は不安定

# 準ニュートン法：Broyden法

ヤコビ行列の近似を反復更新する

- 初期値$\boldsymbol{x}^{(0)}$，初期ヤコビ行列$B^{(0)}$, $k=0$
- 収束するまで以下を反復
    - $\boldsymbol{d}^{(k)} = - (B^{(k)})^{-1} \boldsymbol{f}(\boldsymbol{x}^{(k)})$
        - $B^{(k)} \boldsymbol{d}^{(k)} = -  \boldsymbol{f}(\boldsymbol{x}^{(k)})$を解く
    - 解を更新：$\boldsymbol{x}^{(k+1)} = \boldsymbol{x}^{(k)} +\boldsymbol{d}^{(k)}$
    - 移動量（解の変化量）を計算：$\boldsymbol{s}^{(k)} = \boldsymbol{x}^{(k+1)} - \boldsymbol{x}^{(k)}$
    - 関数値の変化量を計算：$\boldsymbol{y}^{(k)} = \boldsymbol{f}(\boldsymbol{x}^{(k+1)}) - \boldsymbol{f}(\boldsymbol{x}^{(k)})$
    - $B^{(k)}, \boldsymbol{s}^{(k)}, \boldsymbol{y}^{(k)}$を用いて$B^{(k+1)}$を更新




# Broyden法の更新式

\begin{align*}
B^{(k+1)}
&=
B^{(k)} +
\frac{(\boldsymbol{y}^{(k)} - B^{(k)} \boldsymbol{s}^{(k)}) \boldsymbol{s}^{(k)}{}^T}
{\boldsymbol{s}^{(k)}{}^T \boldsymbol{s}^{(k)}}
\\
&=
B^{(k)} +
\frac{\boldsymbol{f}(\boldsymbol{x}^{(k+1)}) \boldsymbol{s}^{(k)}{}^T}
{\boldsymbol{s}^{(k)}{}^T \boldsymbol{s}^{(k)}}
\\
\end{align*}

# 多次元で$n \neq m$の場合

ニュートン法（多次元で $n=m$ の場合の方法）が使えるか？

- $
J(\boldsymbol{x}^{(k)}) \boldsymbol{d}^{(k)}
= \boldsymbol{f}(\boldsymbol{x}^{(k)})
$ を解く

しかし

- ヤコビ行列 $J(\boldsymbol{x}^{(k)})$が正方ではない
  - 逆行列が存在しない $\rightarrow$ 以下の最適化問題に置き換え

$$
\min_{\boldsymbol{x}}
\frac{1}{2} \| \boldsymbol{f}(\boldsymbol{x}) \|^2
$$
または
$$
\min_{\boldsymbol{x}}
\frac{1}{2} \sum_{i=1}^m f_i^2(\boldsymbol{x})
$$


# Gauss-Newton法（GN）

ヘッセ行列$H$を$J^T J$で近似


1. $J^T J  \boldsymbol{d}^{(k)} = - J^T \boldsymbol{f}$を解く
2. $\boldsymbol{x}^{(k+1)} = \boldsymbol{x}^{(k)} + \alpha \boldsymbol{d}^{(k)}$


ここで$J=J(\boldsymbol{x}^{(k)})$，$\boldsymbol{f}=\boldsymbol{f}(\boldsymbol{x}^{(k)})$


# Levenberg-Marquardt法（LM）

ヘッセ行列の近似$J^T J$が正定値になるように$\lambda^{(k)} \boldsymbol{I}$を加える

1. $(J^T J + \lambda^{(k)} \boldsymbol{I}) \boldsymbol{d}^{(k)} = - J^T \boldsymbol{f}$を解く
2. $\boldsymbol{x}^{(k+1)} = \boldsymbol{x}^{(k)} + \alpha \boldsymbol{d}^{(k)}$

$\lambda^{(k)}$は反復ごとに調整される正則化パラメータ

# 多次元で$n \neq m$の場合

ニュートン法（多次元で $n=m$ の場合の方法）が使えるか？

- $
J(\boldsymbol{x}^{(k)}) \boldsymbol{d}^{(k)}
= \boldsymbol{f}(\boldsymbol{x}^{(k)})
$ を解く

しかし

- ヤコビ行列 $J(\boldsymbol{x}^{(k)})$が正方ではない
  - 逆行列が存在しない $\rightarrow$ 一般逆行列（擬似逆行列）を用いる



# 一般逆行列・擬似逆行列 $A^+$

非正方行列$A \in R^{m \times n}$について，連立方程式
$$
A \boldsymbol{x} = \boldsymbol{b}
$$
の最小二乗解または最小ノルム解は
$$
\boldsymbol{x} = A^+ \boldsymbol{b}
$$

ここで一般逆行列・擬似逆行列 $A^+$は以下を満たす

- $A^+ A A^+ = A^+$
- $(A^+ A)^T = A^+ A$
- $(A A^+)^T = A A^+$
- $(A^+)^+ = A$
- $A^+ = A^{-1}$（$A^{-1}$が存在すれば）

# SVDを用いた擬似逆行列の構成


$ A \in R^{m \times n} $ の特異値分解を
$$
A = U W V^T
$$
とすると，
$A$の擬似逆行列は
$$
A^+ = V W^{-1} U^T
$$


# 擬似逆行列を用いたGN

ニュートンステップの連立方程式
$$
J \boldsymbol{d} = - \boldsymbol{f}
$$
を擬似逆行列を用いて解くと
$$
\boldsymbol{d} = - J^+ \boldsymbol{f}
$$

これは$m \neq n$のときGauss−Newton法，$m=n$のときニュートン法になる

- $\boldsymbol{d}^{(k)} = - J(\boldsymbol{x}^{(k)})^+ \boldsymbol{f}(\boldsymbol{x}^{(k)})$
- $\boldsymbol{x}^{(k+1)} = \boldsymbol{x}^{(k)} +\boldsymbol{d}^{(k)}$
