以下は最小ノルム解を導く過程である．次の式展開のステップを順を追って導出せよ．

2. ラグランジュ関数 $ L(x, \lambda) $ を $ x $ および $ \lambda $ について微分せよ。
3. 導出された式を用いて、最小ノルム解 $ x $ を求めよ。
